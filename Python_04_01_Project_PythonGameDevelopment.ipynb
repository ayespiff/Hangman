```
PROGRAM Hangman

    // ---------- Constants ----------
    SET MAX_WRONG_GUESSES TO 6
    SET WORD_BANK TO
        easy   : list of 3-letter words
        medium : list of 6-letter words
        hard   : list of 8 to 10-letter words
    SET HANGMAN_STAGES TO list of 7 gallows drawings (0 to 6 wrong guesses)

    // ---------- Global counters ----------
    SET games_played, games_won, games_lost TO 0
    SET total_guesses, current_streak, best_streak TO 0


    // ---------- Main program ----------
    FUNCTION main()
        CALL display_instructions()
        REPEAT
            CALL play_round()
            CALL show_stats()
        UNTIL play_again() RETURNS False
        DISPLAY "Thanks for playing Hangman. Goodbye!"
    END FUNCTION


    // ---------- One game ----------
    FUNCTION play_round()
        SET difficulty TO choose_difficulty()
        SET word TO random word from WORD_BANK[difficulty]
        SET guessed_letters TO empty set
        SET wrong_guesses TO 0
        SET guesses_used TO 0

        WHILE wrong_guesses < MAX_WRONG_GUESSES
            CALL display_board(word, guessed_letters, wrong_guesses)
            SET guess TO get_valid_guess(guessed_letters)
            ADD guess TO guessed_letters
            INCREMENT guesses_used BY 1

            IF guess IS IN word THEN
                DISPLAY "Yes! The letter is in the word."
                IF every letter of word IS IN guessed_letters THEN
                    DISPLAY "You win!" AND the word
                    CALL update_stats(won = True, guesses_used)
                    RETURN True
                END IF
            ELSE
                INCREMENT wrong_guesses BY 1
                DISPLAY "Sorry, the letter is not in the word."
            END IF
        END WHILE

        // Loop only ends here when all lives are used
        CALL display_board(word, guessed_letters, wrong_guesses)
        DISPLAY "Game over!" AND the word
        CALL update_stats(won = False, guesses_used)
        RETURN False
    END FUNCTION


    // ---------- Input validation ----------
    FUNCTION choose_difficulty()
        LOOP
            INPUT choice
            IF choice IS "1", "2" OR "3" THEN
                RETURN matching level (easy, medium or hard)
            ELSE
                DISPLAY "Invalid choice. Please type 1, 2 or 3."
            END IF
        END LOOP
    END FUNCTION

    FUNCTION get_valid_guess(guessed_letters)
        LOOP
            INPUT guess
            CONVERT guess TO lowercase AND remove spaces
            IF length of guess IS NOT 1 THEN
                DISPLAY "Please enter exactly one letter."
            ELSE IF guess IS NOT a letter from a to z THEN
                DISPLAY "That is not a letter. Use A to Z only."
            ELSE IF guess IS IN guessed_letters THEN
                DISPLAY "You already guessed that letter."
            ELSE
                RETURN guess
            END IF
        END LOOP
    END FUNCTION

    FUNCTION play_again()
        LOOP
            INPUT answer
            IF answer IS "y" OR "yes" THEN RETURN True
            IF answer IS "n" OR "no"  THEN RETURN False
            DISPLAY "Please answer y or n."
        END LOOP
    END FUNCTION


    // ---------- Statistics ----------
    FUNCTION update_stats(won, guesses_used)
        USE global counters
        INCREMENT games_played BY 1
        INCREMENT total_guesses BY guesses_used
        IF won THEN
            INCREMENT games_won BY 1
            INCREMENT current_streak BY 1
            SET best_streak TO larger of best_streak AND current_streak
        ELSE
            INCREMENT games_lost BY 1
            SET current_streak TO 0
        END IF
    END FUNCTION

    FUNCTION show_stats()
        IF games_played > 0 THEN
            SET win_rate TO games_won / games_played × 100
            SET avg_guesses TO total_guesses / games_played
        ELSE
            SET win_rate AND avg_guesses TO 0
        END IF
        DISPLAY all counters, win_rate AND avg_guesses
    END FUNCTION


    // ---------- Display ----------
    FUNCTION display_instructions()
        DISPLAY welcome message AND rules
    END FUNCTION

    FUNCTION display_board(word, guessed_letters, wrong_guesses)
        DISPLAY HANGMAN_STAGES[wrong_guesses]
        DISPLAY word with unguessed letters shown as "_"
        DISPLAY letters guessed that are NOT in word
        DISPLAY lives left (MAX_WRONG_GUESSES - wrong_guesses)
    END FUNCTION


    // ---------- Start ----------
    CALL main()
    IF player presses Ctrl + C THEN
        DISPLAY "Game interrupted." AND exit cleanly
    END IF

END PROGRAM
```

````python
"""
Hangman
=======
A command-line Hangman game built for the Python project.

The computer picks a random word from a themed word bank and the player
guesses one letter at a time. Each wrong guess adds a part to the gallows.
Six wrong guesses and the game is lost. Statistics are tracked across
every game played in the session.

Run with:  python hangman.py
"""

import random
import string

# Constants: fixed values, written in capitals by PEP 8 convention

MAX_WRONG_GUESSES = 6

# Word bank grouped by difficulty level (the "additional levels" feature)
WORD_BANK = {
    "easy": ["cat", "dog", "sun", "map", "cup", "bus", "hat", "pen"],
    "medium": ["python", "planet", "garden", "rocket", "silver", "bridge"],
    "hard": ["algorithm", "variable", "dictionary", "function",
             "iteration", "exception"],
}

# Gallows drawings, indexed by the number of wrong guesses (0 to 6)
HANGMAN_STAGES = [
    """
  +---+
  |   |
      |
      |
      |
      |
=========""",
    """
  +---+
  |   |
  O   |
      |
      |
      |
=========""",
    """
  +---+
  |   |
  O   |
  |   |
      |
      |
=========""",
    """
  +---+
  |   |
  O   |
 /|   |
      |
      |
=========""",
    """
  +---+
  |   |
  O   |
 /|\\  |
      |
      |
=========""",
    """
  +---+
  |   |
  O   |
 /|\\  |
 /    |
      |
=========""",
    """
  +---+
  |   |
  O   |
 /|\\  |
 / \\  |
      |
=========""",
]

# Global counters: live outside every function and track session statistics

games_played = 0
games_won = 0
games_lost = 0
total_guesses = 0
current_streak = 0
best_streak = 0

# Display functions

def display_instructions():
    """Print the welcome message and the rules of the game."""
    print("=" * 50)
    print("               WELCOME TO HANGMAN")
    print("=" * 50)
    print("How to play:")
    print("  1. Choose a difficulty: easy, medium or hard.")
    print("  2. Guess the hidden word one letter at a time.")
    print(f"  3. You lose after {MAX_WRONG_GUESSES} wrong guesses.")
    print("  4. Repeated guesses are not counted against you.")
    print("=" * 50)


def mask_word(word, guessed_letters):
    """
    Return the word with unguessed letters hidden as underscores.

    Args:
        word (str): The secret word.
        guessed_letters (set): Letters the player has guessed so far.

    Returns:
        str: For example 'p _ t h _ n' for 'python' after guessing p, t, h, n.
    """
    return " ".join(letter if letter in guessed_letters else "_"
                    for letter in word)


def display_board(word, guessed_letters, wrong_guesses):
    """
    Print the gallows, the masked word and the letters guessed wrongly.

    Args:
        word (str): The secret word.
        guessed_letters (set): Letters the player has guessed so far.
        wrong_guesses (int): Number of incorrect guesses made.
    """
    wrong_letters = sorted(letter for letter in guessed_letters
                           if letter not in word)

    print(HANGMAN_STAGES[wrong_guesses])
    print(f"\nWord:          {mask_word(word, guessed_letters)}")
    print(f"Wrong letters: {', '.join(wrong_letters) or 'none yet'}")
    print(f"Lives left:    {MAX_WRONG_GUESSES - wrong_guesses}")


def show_stats():
    """Print the session statistics held in the global counters."""
    # Avoid dividing by zero if no games have been played
    win_rate = (games_won / games_played * 100) if games_played else 0
    avg_guesses = (total_guesses / games_played) if games_played else 0

    print("\n" + "-" * 30)
    print("       YOUR STATISTICS")
    print("-" * 30)
    print(f"Games played:     {games_played}")
    print(f"Games won:        {games_won}")
    print(f"Games lost:       {games_lost}")
    print(f"Win rate:         {win_rate:.1f}%")
    print(f"Avg guesses/game: {avg_guesses:.1f}")
    print(f"Current streak:   {current_streak}")
    print(f"Best streak:      {best_streak}")
    print("-" * 30)

# Input functions: every user input is validated before it is used


def choose_difficulty():
    """
    Ask the player to pick a difficulty level until a valid choice is made.

    Returns:
        str: 'easy', 'medium' or 'hard'.
    """
    options = {"1": "easy", "2": "medium", "3": "hard"}

    while True:
        choice = input("\nChoose difficulty (1 = easy, 2 = medium, "
                       "3 = hard): ").strip()
        if choice in options:
            print(f"Difficulty set to {options[choice]}.")
            return options[choice]
        print("Invalid choice. Please type 1, 2 or 3.")


def get_valid_guess(guessed_letters):
    """
    Ask for a letter until the player enters one that is valid and new.

    Rejects empty input, multiple characters, numbers, symbols and
    letters that have already been guessed.

    Args:
        guessed_letters (set): Letters the player has guessed so far.

    Returns:
        str: A single lowercase letter not guessed before.
    """
    while True:
        guess = input("\nGuess a letter: ").strip().lower()

        if len(guess) != 1:
            print("Please enter exactly one letter.")
        elif guess not in string.ascii_lowercase:
            print("That is not a letter. Use A to Z only.")
        elif guess in guessed_letters:
            print(f"You already guessed '{guess}'. Try another letter.")
        else:
            return guess


def play_again():
    """
    Ask whether the player wants another game until they answer y or n.

    Returns:
        bool: True to play again, False to quit.
    """
    while True:
        answer = input("\nPlay again? (y/n): ").strip().lower()
        if answer in ("y", "yes"):
            return True
        if answer in ("n", "no"):
            return False
        print("Please answer y or n.")


# Game logic

def update_stats(won, guesses_used):
    """
    Update the global counters at the end of a game.

    Args:
        won (bool): True if the player won the round.
        guesses_used (int): Number of valid guesses made in the round.
    """
    # 'global' is required to change variables defined outside the function
    global games_played, games_won, games_lost, total_guesses
    global current_streak, best_streak

    games_played += 1
    total_guesses += guesses_used

    if won:
        games_won += 1
        current_streak += 1
        best_streak = max(best_streak, current_streak)
    else:
        games_lost += 1
        current_streak = 0


def play_round():
    """
    Play a single game of Hangman from difficulty choice to win or loss.

    Returns:
        bool: True if the player guessed the word, False otherwise.
    """
    difficulty = choose_difficulty()
    word = random.choice(WORD_BANK[difficulty])
    guessed_letters = set()   # a set stops the same letter being stored twice
    wrong_guesses = 0
    guesses_used = 0

    # Keep playing while the player still has lives left
    while wrong_guesses < MAX_WRONG_GUESSES:
        display_board(word, guessed_letters, wrong_guesses)
        guess = get_valid_guess(guessed_letters)
        guessed_letters.add(guess)
        guesses_used += 1

        if guess in word:
            print(f"Yes! '{guess}' is in the word.")
            # The word is solved once every letter in it has been guessed
            if all(letter in guessed_letters for letter in word):
                display_board(word, guessed_letters, wrong_guesses)
                print(f"\nYou win! The word was '{word}'.")
                update_stats(True, guesses_used)
                return True
        else:
            wrong_guesses += 1
            print(f"Sorry, '{guess}' is not in the word.")

    # The loop only ends here if the player ran out of lives
    display_board(word, guessed_letters, wrong_guesses)
    print(f"\nGame over! The word was '{word}'.")
    update_stats(False, guesses_used)
    return False


def main():
    """Run the game loop until the player chooses to quit."""
    display_instructions()

    try:
        while True:
            play_round()
            show_stats()
            if not play_again():
                break
    except (KeyboardInterrupt, EOFError):
        # Exit cleanly if the player presses Ctrl+C instead of crashing
        print("\n\nGame interrupted.")

    print("\nThanks for playing Hangman. Goodbye!")


# Only start the game when this file is run directly, not when imported
if __name__ == "__main__":
    main()
````
